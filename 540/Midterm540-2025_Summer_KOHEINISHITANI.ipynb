### Wine Dataset Description:
The Wine dataset is a built-in dataset in Scikit-Learn primarily used for classification tasks. It originates from three different types of wine, each described by the results of 13 different chemical analyses. The objective is to predict the origin of the wine based on these chemical attributes.

### Feature Explanations:
- alcohol - Alcohol content
- malic_acid - Malic acid content, a naturally occurring organic acid that is found in various plants and fruits.
- ash - Total ash content, a measure of the inorganic residue remaining after the evaporation of water and the removal of organic material.
- alcalinity_of_ash - Alkalinity of the ash, measuring how basic the ash is.
- magnesium - Magnesium content.
- total_phenols - Total phenols content, compounds which have antioxidant properties.
- flavanoids - Flavanoid content, a group of polyphenols which are known for their health benefits.
- nonflavanoid_phenols - Non-flavanoid phenols content, another type of polyphenol but not classified as flavanoids.
- proanthocyanins - Proanthocyanidin content, polyphenolic antioxidants found in many plants.
- color_intensity - Color intensity, an indicative measure of the wine type and quality.
- hue - Hue, the type of color the wine has.
- od280/od315_of_diluted_wines - Ratio of OD280 to OD315 of diluted wines, a measure often used to gauge the protein content in wine.
- proline - Proline content, an amino acid that is often used as a marker for wine quality.


    - The target variable, named **target**, indicates the origin of the wine, with three possible classes.

### Q2: Data Splitting and Preparation
(5 pts)

Split the Wine dataset into training and test sets and inspect the size of both sets.

-  Allocate 80% of the data for training and 20% for testing.
-  Use `random_state = 7`
-  Display the size of the training and testing data

```python
from sklearn.model_selection import train_test_split

X = df.drop('target', axis=1)
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=7)
```

```python
print("x_training size:", X_train.shape)
print("y_training size:", y_train.shape)
print("x_test size:", X_test.shape)
print("y_test size:", y_test.shape)
```

### Q3: Feature Scaling

(10 pts)

Standardize the features in the training and test sets.

- Use `StandardScaler` from scikit-learn.
- Fit the scaler on the training data and transform both the training and testing data.
- Display the first 5 rows of the scaled training data.

*Note*: All subsequent model building should use the scaled features (e.g., `X_train_scaled` for training and `X_test_scaled` for testing).

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = pd.DataFrame(scaler.transform(X_train), columns=X_train.columns)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns)
```

```python

print("X_train_scaled:")
print(X_train_scaled.head(5))


print("X_test_scaled:")
print(X_test_scaled.head(5))

```

### Q4: Logistic regression model
(20 pts)

##### a)
- Instantiate a logistic regression model.
- Train the model on the training data.
- Make predictions on the training set and calculate the accuracy (optional).
- Make predictions on the test set and calculate the accuracy.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

logisticRegression = LogisticRegression(random_state=7)
logisticRegression.fit(X_train, y_train)

# prediction against training data
y_pred_train = logisticRegression.predict(X_train)
print("Accuracy: ", accuracy_score(y_train, y_pred_train))


# prediction against test data
y_pred_test = logisticRegression.predict(X_test)
print("Accuracy: ", accuracy_score(y_test, y_pred_test))
```

### Q5: Linear Support Vector Machine (SVM) model
(20 pts)

#### a):
- Instantiate a linear SVM model.
- Train the model on training data.
- Calculate accuracy on training and test set.

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score  

svm = LinearSVC(random_state=7)
svm.fit(X_train_scaled, y_train)

y_pred_train = svm.predict(X_train_scaled)
print("Training Accuracy: ", accuracy_score(y_train, y_pred_train))


y_pred_test = svm.predict(X_test_scaled)
print("Test Accuracy: ", accuracy_score(y_test, y_pred_test))

```

#### b). Exploring the Impact of Regularization Parameter C:

- To examine the impact of the regularization strength on the model's training and test accuracy.
- Understand how the choice of C influences the model's generalization capability.

Train a Linear SVM model with different values of C: 0.001, 0.01, 0.1, 1, 10, and 100. Show the training and test accuracies for each value.

```python
c_values = [0.001, 0.01, 0.1, 1, 10, 100]

results = []

for c in c_values:
    svm = LinearSVC(C=c, random_state = 7)
    svm.fit(X_train_scaled, y_train)

    y_pred_test = svm.predict(X_test_scaled)
    y_pred_train = svm.predict(X_train_scaled)

    test_accuracy = accuracy_score(y_test, y_pred_test)
    train_accuracy = accuracy_score(y_train, y_pred_train)
   
    results.append({
        'C': c,
        'Test Accuracy': test_accuracy,
        'Train Accuracy': train_accuracy
    })
    

results_df = pd.DataFrame(results)
print(results_df)
```

**Question: How would you select an optimal value for C based on the training and test accuracies?**

Answer: 
Regarding model evalutation, We should pick up the model with the highest Test accuracy. 
Given that rule, C value 0.001 and 0.01 are the best model among different C values. Smaller C values regularize the model, and larger C values do not regularize the model, so if test accuracy is same, we should pick up the model with smaller C value. Hence, C value 0.001 is the best model.
